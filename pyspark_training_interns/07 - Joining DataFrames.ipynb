# Joining DataFrames

Real datasets often split related information into separate tables. Joins let us combine those tables using a shared key.

---
## Learning objectives

By the end of this notebook, you will be able to:

- identify a shared join key;
- use an inner join to retain matching rows; and
- use a left join to retain every row from the left DataFrame.

---
## Set up related DataFrames

`orders` is transaction data. `customers`, `products`, and `account_managers` are lookup DataFrames that add descriptive information.

```python
orders = spark.createDataFrame( 
    [
        (1001, "C001", "P001", 2, 18.50),
        (1002, "C002", "P002", 1, 750.00),
        (1003, "C001", "P003", 3, 12.00),
        (1004, "C003", "P001", 5, 18.50),
        (1005, "C004", "P004", 2, 85.00),
    ],
    "order_id INT, customer_id STRING, product_id STRING, quantity INT, unit_price DOUBLE",
)

customers = spark.createDataFrame(
    [
        ("C001", "Northwind Supplies", "North"),
        ("C002", "Contoso Retail", "West"),
        ("C003", "Adventure Works", "North"),
        ("C004", "Fabrikam Stores", "South"),
        ("C005", "Wide World Importers", "East"),
    ],
    "customer_id STRING, customer_name STRING, region STRING",
)

products = spark.createDataFrame(
    [
        ("P001", "Notebook", "Stationery"),
        ("P002", "Office Chair", "Furniture"),
        ("P003", "Pen Set", "Stationery"),
        ("P004", "Desk Lamp", "Electronics"),
    ],
    "product_id STRING, product_name STRING, category STRING",
)

account_managers = spark.createDataFrame(
    [
        ("North", "Maya Patel"),
        ("West", "Louis Bernard"),
    ],
    "region STRING, account_manager STRING",
)

orders.show()
customers.show()
products.show()
account_managers.show()
```

---
## Inner join

An inner join keeps only rows whose join key exists in both DataFrames. Here, every order is matched to its customer using `customer_id`.

```python
orders_with_customers = orders.join(
    customers,
    on="customer_id",
    how="inner",
)

orders_with_customers.show()
orders_with_customers.printSchema()
```

---
## Add a second lookup DataFrame

The first join created a new DataFrame, which we can join to `products` using `product_id`. Select only the columns that are useful for the result.

```python
complete_orders = orders_with_customers.join(
    products,
    on="product_id",
    how="inner",
)

complete_orders.select(
    "order_id",
    "customer_name",
    "region",
    "product_name",
    "category",
    "quantity",
).show()
```

---
## Left join

A left join keeps every row from the DataFrame on the left. Customer `C005` has no order, so its order columns appear as `NULL` rather than the customer disappearing from the result.

```python
customers_with_orders = customers.join(
    orders,
    on="customer_id",
    how="left",
)

customers_with_orders.select(
    "customer_id", "customer_name", "region", "order_id"
).show()
```
