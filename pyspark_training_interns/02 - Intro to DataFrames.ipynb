---
## What is a DataFrame?

A DataFrame is a distributed table with named columns and defined data types. It may look familiar if you have used pandas or SQL, but Spark can distribute its rows across partitions when the data is large.

DataFrames are **immutable** (*cannot be changed*): an operation does not change the existing DataFrame. Instead, it describes and returns a new one. We will see this when we use `select()`.

---
## Create a small DataFrame

For this first example, we will create data directly in the notebook. The `StructType` schema makes each column name and data type explicit; later notebooks will read data from files and Lakehouse tables.

```python
employee_rows = [
    (101, "Aisha Khan", "Engineering", 2021),
    (102, "Ben Carter", "Finance", 2019),
    (103, "Chloe Martin", "Engineering", 2023),
    (104, "Daniel Wong", "People", 2020),
    (105, "Elena Rossi", "Finance", 2022),
]

employee_schema = """
    employee_id INT,
    name STRING,
    department STRING,
    start_year INT
"""

employees = spark.createDataFrame(employee_rows, schema=employee_schema)
```

---
## Preview rows

Use `show()` to display a small number of rows. It is useful for a quick check, but it is not how we inspect every row of a large dataset.

```python
employees.show()
```

---
## Inspect the schema

The schema is part of the DataFrame contract: it records the column names, their data types, and whether a value may be missing. Check it before writing transformations, especially when working with data from a file or table.

```python
employees.printSchema()
```

`columns` returns the column names; `dtypes` returns name/type pairs. These are quick ways to inspect DataFrame metadata in Python.

```python
print("Columns:", employees.columns)
print("Data types:", employees.dtypes)
```

---
## Select columns

`select()` creates a new DataFrame containing only the requested columns. The original `employees` DataFrame is unchanged.

When selecting by name, strings are concise and readable.

```python
employee_directory = employees.select("employee_id", "name", "department")
employee_directory.show()
```
