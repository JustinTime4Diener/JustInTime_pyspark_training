## Create a DataFrame

The schema below uses Spark's compact **DDL string** notation. It makes the column names and expected types explicit.

```python
from pyspark.sql import functions as F

product_rows = [
    ('P001', 'Notebook', 'Stationery', 18.50),
    ('P002', 'Office Chair', 'Furniture', 750.00),
    ('P003', 'Pen Set', 'Stationery', 12.00),
    ('P004', 'Desk Lamp', 'Electronics', 85.00),
]

product_schema = '''
    product_id STRING,
    product_name STRING,
    category STRING,
    unit_price DOUBLE
'''

products = spark.createDataFrame(product_rows, schema=product_schema)
```

## Preview and inspect

`show()` previews rows. `printSchema()` displays column names, types, and nullability. `columns` and `dtypes` expose compact metadata in Python.

```python
products.show(truncate=False)
products.printSchema()
print('Columns:', products.columns)
print('Data types:', products.dtypes)
```

## Select and alias columns

`select()` creates a new DataFrame containing the requested expressions. `F.col('column_name')` refers to an existing column when you need to apply an operation such as `alias()`. `alias()` gives a selected expression a useful output name.

```python
product_catalogue = products.select(
    'product_id',
    'product_name',
    F.col('unit_price').alias('list_price'),
)
product_catalogue.show()
```

## DataFrames are immutable

`select()` did not remove columns from `products`. It returned a new DataFrame and left the original definition unchanged.

```python
print('Original:', products.columns)
print('Selected:', product_catalogue.columns)
```
